# Create Field Maps

```python
list_bias = [-1.25, -1 , -0.75 , -0.5, -.375, -.325 , -.3, -0.25, -0.1, -0.05 ,0, +0.05, +0.1, 0.25, 0.5, 0.75, 1., 1.25]
list_stringa = ["PreGravity_20mT_well","GravityRelease_17mT_well","GravityRelease_5mT_well","GravityRelease_2p5mT_well"]

for stringa in list_stringa:
    for bias in list_bias:
        # Parameters
        pm.execute_notebook(
                "Create_Maps.ipynb",
                f"output/Create_Maps{bias}.ipynb",
                parameters={
                    'stringa'           : stringa,
                    'bias'              : bias
                }
            )
```
